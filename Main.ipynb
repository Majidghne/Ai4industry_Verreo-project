# Reconnaissance de logo sur une image de bouchon.

## Import des bibliothèques.

```python
#!pip install tensorflow tensorflow_datasets matplotlib numpy pillow

import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import tensorflow_datasets as tfds

import matplotlib.pyplot as plt

import numpy as np
import os
import PIL
import PIL.Image

print(tf.__version__)
```

## Création, et préparation du `dataset`.

```python
train_images = tf.keras.utils.image_dataset_from_directory(
  "../ai4_workspace/",
  validation_split=0.2,
  subset="training",
  seed=123,
  image_size=(600, 600),
  batch_size=10
)

val_ds = tf.keras.utils.image_dataset_from_directory(
  "../ai4_workspace/",
  validation_split=0.2,
  subset="validation",
  seed=123,
  image_size=(600, 600),
  batch_size=3
)
```

```python
nom_fabricant = [
    "Amorim",
    "CQ_Juan_Costa_Quer",
    "Diam_groupe_Oeneo",
    "JL_Ebrocork",
    "MASilva_Corticas",
    "Nomacorc",
    "TL_Travel_Liege",
    "Trescases"
]
```

## Création de la base convolutive.

```python
model = models.Sequential()
model.add(layers.Conv2D(24, (3, 3), activation='relu', input_shape=(600, 600, 3)))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))

model.summary()

model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10))
```

## Entraînement du modèle.

```python
model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

history = model.fit(
    train_images,
    epochs=5,
    validation_data=val_ds
)
```
